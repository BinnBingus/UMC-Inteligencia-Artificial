Exercício 3

```python
import math # Puxar o valor de pi

print("Cálculo de área de um círculo")
Raio = int(input("Digite o raio do circulo : "))

Area = math.pi * (Raio * Raio)
print("A área do circulo de : ", Area)
```

Exercício 4

```python
print("Conversor de Temperatura de Celsius para Fahrenheit")

Celsius = int(input("Digite o valor da temperatura em Celsius : "))

Fahrenheit = Celsius * 9/5 + 32 # a fórmula de acordo com o enunciado
print("A temperatura em Fahrenheit fica de : ", Fahrenheit)
```

Exercício 5

```python
print("Imprimidor de Tabuada")

Numero = int(input("Digite um número para realizar a Tabuada : "))

for i in range(1, 11): # cria um Loop de acordo com o range que seria a quantidade 10
    resultado = Numero * i
    print(f"{Numero} x {i} = {resultado}") # o print com "f" remove os espaços diferente da virgula, usando o formato f-string
```

```python
print("Programa para exebir Números pares de 1 a 20")

for i in range(1, 21):
  if i % 2 == 0: # eu fiz ele pegar o resto da divisão e ver se é igual a zero, que seria PAR
    print(f"O número {i} é Par")
```
