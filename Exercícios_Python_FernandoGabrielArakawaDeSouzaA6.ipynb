```python
print("Verificador de números primos")

Divisor = 0
Numero = int(input("Digite um número para verificar se ele é primo : "))

for i in range(1, Numero + 1):
  if Numero % i == 0: # Aqui verificamos se resta alguma divisão do número dividido, caso seja 0, ele aumentara o seu divisor
    Divisor += 1

if Divisor == 2: # se o Divisor for igual a 2, ele vai entender que ele é primo, já que pode ser dividio por 1 e ele mesmo
  print("O seu número é primo")
else:
  print("O seu número não é primo")
```

```python
print("Busca binária")

import bisect # Import do bisect, ele ajuda a encontrar a posição de um número dentro da lista, e descobrir o seu indice dentro dela

Lista = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50] # lista genérica
print(Lista)

Numero = int(input("Digite o número binário que quer buscar : "))

Posicao = bisect.bisect_left(Lista, Numero) # o Bisect left, calcula a lista para ver em qual indice ele é encontrado

if Posicao < len(Lista) and Lista[Posicao] == Numero: # Condições para que o número realmente exista dentro dele e que esteja em sua posição certa
    print(f"Encontrado na posição: {Posicao + 1}") # o Indíce do primeiro lugar é 0, mas eu adiciono 1, para mostrar sua posição de uma lista, como se fosse uma fila
else: # caso não cumpra os requisitos
    print("Não encontrado")
```

Exercício 22

```python
print("Criptografia da Crifra de César")

Mensagem = str(input("Digite a mensagem que quer criptografar : "))

alfabeto = "abcdefghijklmnopqrstuvwxyz" # alfabeto para utilizarmos de referência para a função a abaixo

def cifrador(texto, chave): # def é a criação de uma função e dentro dela podemos passar o texto e a chave de quantas letras iremos pular quando chamar a função

  deslocamento = alfabeto[chave:] + alfabeto[:chave] # comando para deslocar o texto de acordo com a chave, utilizando o alfabeto de referência

  tabela = str.maketrans(alfabeto + alfabeto.upper(),deslocamento + deslocamento.upper()) # tabela de substituição para incluir letras maiusculas e minusculas

  return texto.translate(tabela) # colocor o texto inteiro de uma vez

print(f"A sua mensagem ficou como : {cifrador(Mensagem, 3)}")
```

Exercício 23

```python
print("Decifrador da Crifra de César")

Mensagem = str(input("Digite a mensagem que quer decifrar : "))

alfabeto = "abcdefghijklmnopqrstuvwxyz"

def decifrador(texto, chave):

  deslocamento = alfabeto[-chave:] + alfabeto[:-chave] # o exemplo ficou o mesmo que o exercício anterior a única coisa que eu mudei, foi colocando esse "-" na chave, que já resolvemos

  tabela = str.maketrans(alfabeto + alfabeto.upper(),deslocamento + deslocamento.upper())

  return texto.translate(tabela)

print(f"A sua mensagem ficou como : {decifrador(Mensagem, 3)}")
```

Exercício 24

```python
print("Validador de CPF Brasileiro")

CPF = str(input("Digite um CPF : "))

def ValidarCPF(cpf): # criação da função para chamar ela e realizar a validação
  cpf = "".join(char for char in cpf if char.isdigit()) # Deixar apenas os números dentro do cpf

  if len(cpf) != 11 or len(set(cpf)) == 1: # verificação pra ver se todos os dígitos são iguais ou se não tiver 11 dígitos
    return False

  # validação para os dois dígitos finais de uma vez
  a1 = sum(int(cpf[i]) * (10 - i) for i in range(9)) * 10 % 11 % 10
  a2 = sum(int(cpf[i]) * (11 - i) for i in range(10)) * 10 % 11 % 10
  # essa é a parte importante da validação do CPF, porque os dois ultímos números, são oque deixam ele válido através dessa calculo

  return int(cpf[9]) == a1 and int(cpf[10]) == a2
if ValidarCPF(CPF):
  print("CPF Válido")
else:
  print("CPF Inválido")
```

Exercício 25

```python
print("Categorizar números positivos,negativos e zeros")

Lista = [0, -25, 5, 35, 50, -100, 0] # lista genérica

Positivos = []
Negativos = []
Zeros = []
# listas vazias para depois serem preenchidas com os números

for i in range(len(Lista)): # cria um loop entre os indices para verificar cada número
  if Lista[i] > 0:
    Positivos.append(Lista[i]) # append é usado para adicionar o número dentro de uma lista
  elif Lista[i] < 0:
    Negativos.append(Lista[i])
  else: # caso nenhuma condição seja cumprida, ele vai ser preenchida em zero
    Zeros.append(Lista[i])

print(f"Positivos : {Positivos}")
print(f"Negativos : {Negativos}")
print(f"Zeros : {Zeros}")
```
