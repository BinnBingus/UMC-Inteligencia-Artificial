Exercício 18 - Número Primo



```python
print("Verificador de números primos")

Divisor = 0
Numero = int(input("Digite um número para verificar se ele é primo : "))

for i in range(1, Numero + 1):
  if Numero % i == 0: # Aqui verificamos se resta alguma divisão do número dividido, caso seja 0, ele aumentara o seu divisor
    Divisor += 1

if Divisor == 2: # se o Divisor for igual a 2, ele vai entender que ele é primo, já que pode ser dividio por 1 e ele mesmo
  print("O seu número é primo")
else:
  print("O seu número não é primo")
```

Justificação : Primeiro nós criamos uma variável de divisores que seria a quantidade de divisões feitas, depois pegamos o número que o usuário quer verficar, e assim realizamos um loop com a quantidade de vezes com o número digitado, e depois nós fazemos duas condições que seria se ele é divisível por 0, caso ele seja vamos aumentar a quantidade de Divisor em +1, e depois caso o número de divisores seja igual a 2, nós vamos detectar que ele é um número primo, caso contrário ele não é primo

Exercício 20 - Matriz Transposta

```python
print("Criação de uma Matriz Transposta")

matriz = [
    [1, 2, 3],
    [4, 5, 6]
]

linhas = len(matriz) # descobre o número de linhas e colunas originais
colunas = len(matriz[0])

transposta = []

# cria uma matriz vazia com o tamanho invertido
for i in range(colunas):
    transposta.append([0] * linhas)

for a in range(linhas): # aqui ele preenche a matriz transposta trocando as suas posições
    for b in range(colunas):
        transposta[b][a] = matriz[a][b]

print("Matriz Transposta:")
for linha in transposta:
    print(linha)
```

Justificação : No código ele transforma as linhas de uma matriz original em colunas na nova matriz. Para alcançar isso, o algoritmo primeiro mede as dimensões da tabela inicial e constrói uma nova estrutura vazia com o tamanho invertido, o programa percorre cada elemento usando laços de repetição e inverte os índices de posição de cada número. Esse processo rearranja os dados de forma organizada, convertendo com sucesso uma matriz de 2x3 em uma nova versão de 3x2, no código podemos ver comandos como o ".append()" que serve justamente para adicionar elementos a uma tabela

Exercício 22 - Criptografia de Cifra de César

```python
print("Criptografia da Crifra de César")

Mensagem = str(input("Digite a mensagem que quer criptografar : "))

alfabeto = "abcdefghijklmnopqrstuvwxyz" # alfabeto para utilizarmos de referência para a função a abaixo

def cifrador(texto, chave): # def é a criação de uma função e dentro dela podemos passar o texto e a chave de quantas letras iremos pular quando chamar a função

  deslocamento = alfabeto[chave:] + alfabeto[:chave] # comando para deslocar o texto de acordo com a chave, utilizando o alfabeto de referência

  tabela = str.maketrans(alfabeto + alfabeto.upper(),deslocamento + deslocamento.upper()) # tabela de substituição para incluir letras maiusculas e minusculas

  return texto.translate(tabela) # colocor o texto inteiro de uma vez

print(f"A sua mensagem ficou como : {cifrador(Mensagem, 3)}")
```

Justificação : É solicitado para criarmos uma função de criptografia, logo então fazemos um "def" que tem sua função de receber o texto fornecido pelo usuário para criptografia e quantidade de chaves que vamos pular do alfabeto, criamos uma variável chamada "alfabeto" com todas as letras que podemos usar para a cifra de césar, e dentro da função utilizamos o operador slice novamente para realização do chaveamento, e depois usamos o "str.maketrans()" que serve justamente para substituirmos as letras do alfabeto pelo descolento novo q fizemos, e depois realizamos um "return" que seria parte da "def" aonde retornamos algum resultado, que seria o "texto.translate()" que retorna o texto inteiro junto

Exercício 23 - Decifrador de Cifra de César

```python
print("Decifrador da Cifra de César")

Mensagem = str(input("Digite a mensagem que quer decifrar : "))

alfabeto = "abcdefghijklmnopqrstuvwxyz"

def decifrador(texto, chave):

  deslocamento = alfabeto[-chave:] + alfabeto[:-chave] # o exemplo ficou o mesmo que o exercício anterior a única coisa que eu mudei, foi colocando esse "-" na chave, que já resolvemos

  tabela = str.maketrans(alfabeto + alfabeto.upper(),deslocamento + deslocamento.upper())

  return texto.translate(tabela)

print(f"A sua mensagem ficou como : {decifrador(Mensagem, 3)}")
```

Justificação : O código é o mesmo que o do exercício anterior a única coisa que alteramos é como a criptografia é feita dentro da função, trocamos o numero positivo do operado slice para negativo, assim conseguimos fazer uma decifrador da Cifra de César

Exercício 24 - Validação CPF

```python
print("Validador de CPF Brasileiro")

CPF = str(input("Digite um CPF : "))

def ValidarCPF(cpf): # criação da função para chamar ela e realizar a validação
  cpf = "".join(char for char in cpf if char.isdigit()) # Deixar apenas os números dentro do cpf

  if len(cpf) != 11 or len(set(cpf)) == 1: # verificação pra ver se todos os dígitos são iguais ou se não tiver 11 dígitos
    return False

  # validação para os dois dígitos finais de uma vez
  a1 = sum(int(cpf[i]) * (10 - i) for i in range(9)) * 10 % 11 % 10
  a2 = sum(int(cpf[i]) * (11 - i) for i in range(10)) * 10 % 11 % 10
  # essa é a parte importante da validação do CPF, porque os dois ultímos números, são oque deixam ele válido através dessa calculo

  return int(cpf[9]) == a1 and int(cpf[10]) == a2
if ValidarCPF(CPF):
  print("CPF Válido")
else:
  print("CPF Inválido")
```

Justificação : Após o usuário digitar o seu CPF ele passa para uma função de validar ele, essa função possui um calculo expecífico para calcular o CPF mas em resumo ele calcula os 9 primeiros números do CPF e ocorre uma validação nos ultimos dois números dele, o resultadado da multiplicação dos 9 números tem que ser válida com  os ultimos dois números do seu CPF, podemos ver alguns comandos diferentes como o "join", que tem sua função de juntar os elementos que são separados por uma virgula em uma lista, e o "isdigit()" que é pra ver se o elemento fornecido é um número

Exercício 25 - Classificação de Números

```python
print("Categorizar números positivos,negativos e zeros")

Lista = [0, -25, 5, 35, 50, -100, 0] # lista genérica

Positivos = []
Negativos = []
Zeros = []
# listas vazias para depois serem preenchidas com os números

for numero in Lista: # cria um loop entre os indices para verificar cada número
  if numero > 0:
    Positivos.append(numero) # append é usado para adicionar o número dentro de uma lista
  elif numero < 0:
    Negativos.append(numero)
  else: # caso nenhuma condição seja cumprida, ele vai ser preenchida em zero
    Zeros.append(numero)

print(f"Positivos : {Positivos}")
print(f"Negativos : {Negativos}")
print(f"Zeros : {Zeros}")
```

Justificação : Primeiro criamos as 3 listas com a categoria de seus nomes e depois realizamos um loop dentro dela e criamos suas condições expecificamos para separar cada numero como por exemplo se ele for acima de 0 é positivo e se for abaixo é negativo, caso não seja os dois ele vai entrar na lista dos zeros

Exercício 26 - Calculadora Simples

```python
print("Calculadora Simples")

print("-=-=-=-=-=-=-= Calculadora -=-=-=-=-=-=-=-")
print("( 1 ) - Adição")
print("( 2 ) - Subtração")
print("( 3 ) - Multiplicação")
print("( 4 ) - Divisão")
print("-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-")

Equacao = int(input("Escolha sua opção : "))

if Equacao > 0 and Equacao < 5: # Verificar se a opção digitada existe
  Numero1 = int(input("Digite o primeiro número : "))
  Numero2 = int(input("Digite o segundo número : "))

  Resp = 0

  if Equacao == 1: # Condições para realizar seus calculos com a opção desejada
    Resp = Numero1 + Numero2
  elif Equacao == 2:
    Resp = Numero1 - Numero2
  elif Equacao == 3:
    Resp = Numero1 * Numero2
  else:
    if Numero1 == 0 or Numero2 == 0: # Verificar se algum número foi zero, pois não existe divisão por 0
      print("Não existe divisão por ZERO")
    else:
      Resp = Numero1 / Numero2

  print(f"Resultado : {Resp}")
else:
  print("Sua opção não existe")
```

Exercício 30 - Jogo da Forca

```python
print("Jogo da Forca")

import random

palavras = ["Fabiano", "Inteligência", "Jogo", "Python", "Programação", "Trabalho", "Biologia"] # lista das palavras
palavraSecreta = random.choice(palavras).upper() # pega uma palavra aleatória em uppercase

letrasAcertadas = [] # criação das listas de acertos e erros
letrasErradas = []
tentativasRestantes = 6 # tentativas

while tentativasRestantes > 0: # enquanto ele tiver tentativas restantes ele vai poder jogar
    status_palavra = ""
    for letra in palavraSecreta: # transformar as letras da palavra secreta em "_" para iniciar o jogo
        if letra in letrasAcertadas:
            status_palavra += letra + " "
        else:
            status_palavra += "_ "
    print(f"\n-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-")
    print(f"\nPalavra: {status_palavra}")

    if "_" not in status_palavra: # caso não tenho mais letras invisíveis ele mostra que vc ganhou!
        print("AEEEEE tu ganhou!")
        break # quebrar o loop

    print(f"Tentativas restantes: {tentativasRestantes}")
    print(f"Letras erradas: {', '.join(letrasErradas)}")

    chute = input("Digite uma letra: ").upper().strip() # pegar a letra

    if len(chute) != 1 or not chute.isalpha(): # ver se é apenas umas letra e ele estiver no alfabeto
        print("Mano, apenas digite uma letra")
        continue

    if chute in letrasAcertadas or chute in letrasErradas: # verificar se a letra já foi digitada
        print("Essa letra tu já digitou meu fi")
        continue
    if chute in palavraSecreta: # ver se o acerto corresponde a palavra digitada
        print("Agora tu pensou, acertou parabéns")
        letrasAcertadas.append(chute)
    else: # caso não cumpra vai mostrar o erro e adicionar o erro a tabela de tentativas e letras erradas
        print("ERROUUU!!! não é essa letra")
        letrasErradas.append(chute)
        tentativasRestantes -= 1
if tentativasRestantes == 0: # caso acabe suas tentativas é game ove
    print(f"KKKKK muito ruim a palavra era {palavraSecreta}")
```

Justificação : Utilizamos o "While" que ele é um loop que fica rodando até a condição posta do elemento ser cumprida como por exemplo as tentativas, ele vai ficar rodando o código até parar de cumprir esse critério, mas também caso ele ganhe podemos usar o "break" que serve como uma quebra do loop, podemos visualizar comandos como o "continue" que serve caso já cumpra uma certa condição ele vai prosseguir o restante do código, e também o "not" que seria o inverso de uma condição que colocamos tipo "if not" seria uma condição de caso não.

Exercício 31 - Ordenação por Bolha

```python
print("Ordenação por Bolha")

numeros = [36, 67, 1, 747, -36, 15, 48]

tamanho = len(numeros)

for i in range(tamanho): # vai verificar o tamanho da lista e ordenar elas de lado a lado, comparando o número anterior e o próximo número
    for j in range(tamanho - 1):
        if numeros[j] > numeros[j + 1]:
            numeros[j], numeros[j + 1] = numeros[j + 1], numeros[j]

print(f"A ordem ficou como : {numeros}")
```

Justificação : É realizado um loop dentro de um loop, isso serve para a gente comparar os tamanhos entre os outros números, que é justamente oque a ordenação por bolha faz, comparar o tamanho dos números com os outros números e ordenar lado a lado

Exercício 37 - Funções Recursivas

```python
print("Função recursiva para calcular fatorial de um número")

def Fatorial(n): # a função Fatorial permite ficar chamando ela mesma para chegar no resultado, isso é a formula de um calculo fatorial deve ser
  if n <= 1:
    return 1
  return n * Fatorial(n - 1)

Numero = int(input("Digite um número para calcular a sua fatorial : ")) # pegar o número desejado

print(f"O fatorial do seu número é : {Fatorial(Numero)}") # mostrar resultado e a função loopar ele até achar o numero fatorial na função acima
```

Exercício 46 - Calcular Média e Desvio Padrão

```python
print("Calculo de Média e Desvio Padrão")

Lista = [1, 4, 6, 4, 7, 10, 13, 14, 19, 22] # lista genérica

Media = sum(Lista) / len(Lista) # Calculo da Média
print(f"A média ficou como {Media}")

Quadrados = sum((x - Media) ** 2 for x in Lista) # Calculo em quadrado para calcular a variância

Variancia = Quadrados / (len(Lista) - 1) # formula para o Desvio padrão
Desviopadrao = Variancia ** 0.5 # resultado
print(f"o Desvio Padrão ficou como : {Desviopadrao}")
```

Justificação  : Nessa lista podemos calcular a média dela utilizando o "sum()" e o comando "len()" que seria a adição dividido pela quantidade e em seguida utilizamos as fórmulas do quadrado, variância e Desvio Padrão para chegar em seu resultado, quando há "**" significa que é elevado

Exercício 48 - Ordenação por Inserção

```python
print("Ordenação por Inserção dentro de lista")

ListaBaguncada = [6, 1, 3, 2, 5, 7, 9, 8] # lista genérica

def InsertionSort(Lista): # Função para a organização da lista
  for numero in range(1, len(Lista)):
    chave = Lista[numero] # pegar a posição do numero dentro da lista
    i = numero - 1 # Numero anterior

    while i >= 0 and chave < Lista[i]: # comparar com os números e organizar ela dentro da lista
      Lista[i + 1] = Lista[i]
      i -= 1

    Lista[i + 1] = chave # aumentar a sua posição dentro da lista

  return Lista

print(f"A sua lista ficou como : {InsertionSort(ListaBaguncada)}")
```
